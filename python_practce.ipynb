```python
## only positonal args '/'
def sum(a,b,/):
    return a+b

print(sum(10,20))
```

```python
# here is an example of a function that only accepts positional arguments. The '/' in the function definition indicates that all parameters before it must be specified positionally, and cannot be passed as keyword arguments.
def sum(a,b,/):
    return a+b
print(sum(a=10,b=20))
```
